```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()
```

```python

file_name = next(iter(uploaded))

df = pd.read_excel(file_name)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python

print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst 5 rows:")
display(df.head())

print("\nStatistical summary:")
display(df.describe())
```

```python

print("Actual null values in each column:")
print(df.isnull().sum())

print("\nNumber of -200 values in each column:")
print((df == -200).sum())
```

```python
# Convert the -200 marker to NaN in a separate copy for analysis
eda_df = df.copy()

measurement_cols = eda_df.select_dtypes(include=np.number).columns
eda_df[measurement_cols] = eda_df[measurement_cols].replace(-200, np.nan)

missing_percent = (eda_df.isnull().mean() * 100).sort_values(ascending=False)

plt.figure(figsize=(12, 6))
missing_percent.plot(kind="bar")
plt.title("Percentage of Missing Values in Each Column")
plt.ylabel("Missing Values (%)")
plt.xlabel("Columns")
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()

display(missing_percent.to_frame("Missing Percentage"))
```

```python
co_data = df["CO(GT)"].replace(-200, np.nan).dropna()

plt.figure(figsize=(10, 5))
sns.histplot(co_data, kde=True)
plt.title("Distribution of CO(GT) Measurements")
plt.xlabel("CO concentration (mg/m³)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("Valid CO measurements:", len(co_data))
print("Mean CO concentration:", round(co_data.mean(), 3))
print("Median CO concentration:", round(co_data.median(), 3))
```

```python
time_df = df[["Date", "Time", "CO(GT)"]].copy()

time_df["CO(GT)"] = time_df["CO(GT)"].replace(-200, np.nan)

time_df["Timestamp"] = pd.to_datetime(
    time_df["Date"].dt.strftime("%Y-%m-%d") + " " +
    time_df["Time"].astype(str),
    errors="coerce"
)

time_df = time_df.sort_values("Timestamp")

plt.figure(figsize=(14, 5))
plt.plot(time_df["Timestamp"], time_df["CO(GT)"], linewidth=0.7)
plt.title("Hourly CO Concentration Over Time")
plt.xlabel("Date")
plt.ylabel("CO concentration (mg/m³)")
plt.tight_layout()
plt.show()
```

```python

time_df = df[["Date", "Time", "CO(GT)"]].copy()

time_df["CO(GT)"] = time_df["CO(GT)"].replace(-200, np.nan)

time_df["Date"] = pd.to_datetime(
    time_df["Date"], dayfirst=True, errors="coerce"
)

time_df["Timestamp"] = pd.to_datetime(
    time_df["Date"].dt.strftime("%Y-%m-%d") + " " +
    time_df["Time"].astype(str).str.strip(),
    errors="coerce"
)

time_df = time_df.dropna(subset=["Timestamp"])
time_df = time_df.sort_values("Timestamp")

plt.figure(figsize=(14, 5))
plt.plot(time_df["Timestamp"], time_df["CO(GT)"], linewidth=0.7)
plt.title("Hourly CO Concentration Over Time")
plt.xlabel("Date")
plt.ylabel("CO concentration (mg/m³)")
plt.tight_layout()
plt.show()
```

```python
# Create a copy of the original dataset
clean_df = df.copy()

# Replace -200 with NaN in numerical measurement columns
measurement_cols = clean_df.select_dtypes(include=np.number).columns
clean_df[measurement_cols] = clean_df[measurement_cols].replace(-200, np.nan)

# Convert Date into datetime format
clean_df["Date"] = pd.to_datetime(
    clean_df["Date"], dayfirst=True, errors="coerce"
)

# Create a timestamp using Date and Time
clean_df["Timestamp"] = pd.to_datetime(
    clean_df["Date"].dt.strftime("%Y-%m-%d") + " " +
    clean_df["Time"].astype(str).str.strip(),
    errors="coerce"
)

# Sort records chronologically
clean_df = clean_df.dropna(subset=["Timestamp"])
clean_df = clean_df.sort_values("Timestamp").reset_index(drop=True)

print("Data cleaning started successfully!")
print("Shape:", clean_df.shape)
display(clean_df.head())
```

```python
print("Missing values in each column:")
display(clean_df.isnull().sum().to_frame("Missing Count"))

print("\nMissing percentage in each column:")
display(
    (clean_df.isnull().mean() * 100)
    .round(2)
    .to_frame("Missing Percentage")
)
```

```python
# Keep only rows with a valid CO measurement for the target
clean_df["Target_CO_Next_Hour"] = clean_df["CO(GT)"].shift(-1)

# Remove the final row because it has no next-hour target
model_df = clean_df.dropna(
    subset=["CO(GT)", "Target_CO_Next_Hour"]
).copy()

print("Dataset prepared for next-hour prediction.")
print("Shape:", model_df.shape)

display(
    model_df[
        ["Timestamp", "CO(GT)", "Target_CO_Next_Hour"]
    ].head(10)
)
```
